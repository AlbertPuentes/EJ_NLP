**1. Normalización de texto**

Descripción: Proceso de estandarización del texto para reducir la variabilidad del vocabulario. Incluye convertir a minúsculas, eliminar caracteres especiales, acentos y espacios redundantes.

```python
import re
import unicodedata

def normalizar_texto(texto: str) -> str:
    # Convertir a minúsculas
    texto = texto.lower()
    # Eliminar acentos
    texto = ''.join(c for c in unicodedata.normalize('NFD', texto) if unicodedata.category(c) != 'Mn')
    # Eliminar puntuación y caracteres especiales
    texto = re.sub(r'[^a-z0-9\s]', '', texto)
    # Limpiar espacios en blanco extra
    return re.sub(r'\s+', ' ', texto).strip()

texto = "¡Prueba! De NormAlización, CON ÁcentOs y siGnos de Puntuación..."
print(normalizar_texto(texto))
```

**2. Stemming**

Descripción: Algoritmo heurístico que recorta los sufijos de las palabras para reducirlas a su raíz o "stem" morfológico, sin considerar el contexto léxico

```python
import nltk
from nltk.stem.snowball import SnowballStemmer

stemmer = SnowballStemmer('spanish')
palabras = ["corriendo", "corredor", "corrió", "procesamiento", "procesadores"]

stems = {palabra: stemmer.stem(palabra) for palabra in palabras}
print(stems)
```

**3. Stopwords**

Descripción: Filtrado de palabras frecuentes (artículos, preposiciones, conjunciones) que aportan poco valor semántico en el análisis de texto.

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Descarga el tokenizador de palabras vacías y punkt.
nltk.download('stopwords')
nltk.download('punkt')

# Descarga 'punkt_tab' para la tokenización de idiomas específicos
nltk.download('punkt_tab')

texto = "El procesamiento de lenguaje natural es una disciplina fundamental para la inteligencia artificial."

# Especifique el idioma 'español' para que word_tokenize se alinee con las palabras vacías en español.
palabras = word_tokenize(texto.lower(), language='spanish')
stop_words = set(stopwords.words('spanish'))

filtrado = [w for w in palabras if w not in stop_words and w.isalnum()]
print(filtrado)
```

**4. Term Frequency (TF)**

Descripción: Medida de la frecuencia con la que aparece un término dentro de un documento individual (frecuencia absoluta o relativa).

```python
from collections import Counter
import re

texto = "el gato come pescado y el perro come carne"
palabras = re.findall(r'\b\w+\b', texto.lower())

tf_counts = Counter(palabras)
total_palabras = len(palabras)
tf_relativo = {word: count / total_palabras for word, count in tf_counts.items()}

print(tf_relativo)
```

**5. Inverse Document Frequency (IDF)**

Descripción: Métrica que evalúa la importancia de una palabra penalizando los términos que aparecen con mucha frecuencia en todo el corpus de documentos.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documentos = [
    "El procesamiento de lenguaje natural permite procesar grandes cantidades de informacion",
    "Los modelos analizan oraciones complejas para extraer su intención, contexto o tono.",
    "la inteligencia artificial utiliza modelos de lenguaje"
]

vectorizer = TfidfVectorizer()
vectorizer.fit(documentos)

for word, idx in vectorizer.vocabulary_.items():
    print(f"Palabra: {word:15s} | IDF: {vectorizer.idf_[idx]:.4f}")
```

**7. Lematización**

Descripción: Proceso morfológico que reduce cada palabra a su forma canónica o lema de diccionario utilizando análisis léxico y gramatical.

```python
import spacy

nlp = spacy.load("es_core_news_sm")
texto = "Los gatos estaban corriendo por los parques mientras jugaban"
doc = nlp(texto)

lemas = [(token.text, token.lemma_) for token in doc]
print(lemas)
```

**11. Question Answering (QA)**

Descripción: Sistema que extrae o genera la respuesta exacta a una pregunta planteada basándose en un contexto o documento dado.

```python
import sys
from transformers import pipeline

def resolver_qa():
    try:
        # Carga del pipeline especificado para español (SQuAD v2)
        qa_pipeline = pipeline(
            "question-answering",
            model="mrm8488/bert-base-spanish-wwm-cased-finetuned-spa-squad2-es"
        )

        contexto = "Python es un lenguaje de programación multiparadigma creado por Guido van Rossum a finales de los 80."
        pregunta = "¿Quién creó el lenguaje Python?"

        # Invocación de la inferencia
        respuesta = qa_pipeline(question=pregunta, context=contexto)

        print("--- RESULTADO DEL PROCESAMIENTO QA ---")
        print(f"Pregunta:   {pregunta}")
        print(f"Respuesta:  {respuesta['answer']}")
        print(f"Confianza:  {respuesta['score']:.4f}")
        print(f"Índices:    Inicio {respuesta['start']} -> Fin {respuesta['end']}")

    except Exception as e:
        print(f"Ejemplo de error durante la ejecución del modelo QA: {e}", file=sys.stderr)

if __name__ == "__main__":
    resolver_qa()
```

**12. Summarization (Resumen de texto)**

Descripción: Condensación de un documento extenso en una representación más breve manteniendo las ideas principales.

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("facebook/bart-large-cnn")
model = AutoModelForSeq2SeqLM.from_pretrained("facebook/bart-large-cnn")

texto_largo = """
El procesamiento del lenguaje natural (PLN) es una rama de la inteligencia artificial que ayuda a las computadoras
a entender, interpretar y manipular el lenguaje humano. El PLN se nutre de muchas disciplinas, entre ellas la
ciencia de la computación y la lingüística computacional, para llenar el vacío entre la comunicación humana
y la comprensión de las computadoras.
"""

inputs = tokenizer(texto_largo, max_length=1024, truncation=True, return_tensors="pt")
summary_ids = model.generate(inputs["input_ids"], max_length=40, min_length=15, do_sample=False)
resumen = tokenizer.decode(summary_ids[0], skip_special_tokens=True)
```

**13. Sentence Similarity (Similitud de oraciones)**

Descripción: Cálculo de la distancia o grado de parecido semántico entre oraciones mediante representaciones en espacios vectoriales (embeddings) y similitud de coseno.

```python
from sentence_transformers import SentenceTransformer, util

model = SentenceTransformer('all-MiniLM-L6-v2')

oracion1 = "El perro corre alegremente en el parque"
oracion2 = "Un canino está corriendo por el jardín"

emb1 = model.encode(oracion1, convert_to_tensor=True)
emb2 = model.encode(oracion2, convert_to_tensor=True)

similitud = util.cos_sim(emb1, emb2)
print(f"Similitud semántica: {similitud.item():.4f}")
```

**15. Translation**

Descripción: Conversión automática de un texto escrito en un idioma de origen a un idioma de destino preservando el significado.

Python

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("Helsinki-NLP/opus-mt-es-en")
model = AutoModelForSeq2SeqLM.from_pretrained("Helsinki-NLP/opus-mt-es-en")

def translate_text(text):
    inputs = tokenizer(text, return_tensors="pt")
    outputs = model.generate(**inputs)
    return tokenizer.decode(outputs[0], skip_special_tokens=True)

texto_es = "El procesamiento de texto facilita el análisis de grandes volúmenes de información."

resultado = translate_text(texto_es)
print(resultado)
```

**16. Text Generation (Basado en cadenas de Markov)**

Descripción: Generación probabilística de secuencias de texto fundamentada en las frecuencias de transición entre palabras consecutivas (N-gramas).

```python
import random

texto_entrenamiento = "el gato come pescado el gato duerme en el sofa el perro come carne"
palabras = texto_entrenamiento.split()

# Construcción de la tabla de transiciones (Bigramas)
bigramas = {}
for i in range(len(palabras) - 1):
    curr, nxt = palabras[i], palabras[i + 1]
    bigramas.setdefault(curr, []).append(nxt)

# Generación de la secuencia
palabra_actual = "el"
generado = [palabra_actual]
for _ in range(7):
    siguientes = bigramas.get(palabra_actual, palabras)
    palabra_actual = random.choice(siguientes)
    generado.append(palabra_actual)

print("Texto generado:", " ".join(generado))
```

**17. Text Mining (Modelado de temas - LDA)**

Descripción: Descubrimiento de estructuras conceptuales y patrones latentes en conjuntos de texto sin etiquetar mediante técnicas como Latent Dirichlet Allocation (LDA).

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation

corpus = [
    "el banco otorga créditos y préstamos con tasa de interés baja",
    "la cuenta bancaria permite transferencias y depósitos financieros",
    "el jugador de fútbol anotó un gol en el partido del campeonato",
    "el equipo de baloncesto ganó la final del torneo con gran ventaja"
]

vectorizer = CountVectorizer(stop_words=['el', 'y', 'con', 'de', 'la', 'del', 'un'])
X = vectorizer.fit_transform(corpus)

lda = LatentDirichletAllocation(n_components=2, random_state=42)
lda.fit(X)

vocabulario = vectorizer.get_feature_names_out()
for idx, topic in enumerate(lda.components_):
    top_words = [vocabulario[i] for i in topic.argsort()[:-4:-1]]
    print(f"Tema {idx + 1}: {', '.join(top_words)}")
```
